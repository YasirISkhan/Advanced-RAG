```python
from langchain_text_splitters import CharacterTextSplitter
```

```python
text = """
In the rapidly evolving domain of artificial intelligence, retrieval-augmented generation architectures have become essential for bridging the knowledge gap between pre-trained foundation models and private enterprise data stores. By combining dense vector embeddings with efficient similarity search engines—such as ChromaDB, Pinecone, or FAISS—developers can query vast document repositories in real time. However, the efficacy of any retrieval pipeline relies fundamentally on the underlying document preparation phase. If source files are segmented arbitrarily, critical context gets severed across chunk boundaries, resulting in hallucinated answers or irrelevant search matches. Consequently, engineers employ sophisticated text-splitting heuristics that respect structural hierarchies like Markdown headers, sentence punctuation, and code block delimiters. Recursive splitting algorithms, for instance, iteratively parse text through a prioritized list of separators until each resulting chunk fits strictly within the specified token or character constraints, thereby maintaining local coherence while maximizing information density for downstream embedding models.
"""
```

```python
splitter = CharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap=0,
    length_function=len,
    separator=""
)
```

```python
chunks = splitter.split_text(text)

print(chunks)
```

```python
print(len(chunks))
```

```python
from termcolor import COLORS, colored
from random import choice
```

```python
print(COLORS)
```

```python
colors_list = list(COLORS.keys())[2:8]

colors_list
```

```python
def display_chunks(chunks):
    colors_list = list(COLORS.keys())[2:8]
    print(f"Total Number of chunks: {len(chunks)}")
    for num, chunk in enumerate(chunks,1):
        print(f"Chunk {num}: Length {len(chunk)} chars")
        print(colored(text=chunk, color=choice(colors_list)), end="\n\n")
```

```python
display_chunks(chunks)
```

```python
def create_chunks(text: str, chunk_size: int, separator: str, chunk_overlap:int =0) -> list[str]:
    splitter = CharacterTextSplitter(
        chunk_size = chunk_size,
        chunk_overlap=chunk_overlap,
        separator=separator
    )

    chunks = splitter.split_text(text=text)
    return chunks
```

```python
display_chunks(create_chunks(text, 100, ""))
```

```python
display_chunks(create_chunks(text, 300, ""))
```

```python
display_chunks(create_chunks(text, 100, " "))
```

```python
display_chunks(create_chunks(text, 100, "\n\n"))
```

# Splitting Documents

```python
from langchain_core.documents import Document
docs = [Document(
    page_content = text, 
    metadata = {'source': "Text on AI"}
)]

print(docs)
```

```python
splitter = CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    separator=""
)
```

```python
chunks = splitter.split_documents(docs)

print(chunks)
```
