```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = """
Hi how are you
My name is Yasir
I am learning RAG
We are practicing RAG
"""
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=10,
    chunk_overlap=0, 
    separators="\n\n"
)
```

```python
splitter.split_text(text)
```
